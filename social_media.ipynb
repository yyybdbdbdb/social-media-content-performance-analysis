```python
df_clean = df.copy()

# 1. Convert engagement rates from text percentages to numeric percentages
rate_columns = [
    'Engagement_Rate',
    'Recalculated_Engagement_Rate'
]

for col in rate_columns:
    df_clean[col + '_Num'] = pd.to_numeric(
        df_clean[col].str.replace('%', '', regex=False),
        errors='coerce'
    )

```

```python
# 2. Standardize CTR to percentage values
def convert_ctr(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value.endswith('%'):
        return pd.to_numeric(value[:-1], errors='coerce')

    # Decimal ratios such as 0.0108 become 1.08%
    return pd.to_numeric(value, errors='coerce') * 100

df_clean['CTR_Percentage'] = df_clean['Click_Through_Rate'].apply(convert_ctr)

```

```python
# 3. Convert Post_Date to a proper date
df_clean['Post_Date_Clean'] = pd.to_datetime(
    df_clean['Post_Date'].str[:10],
    format='%d-%m-%Y',
    errors='coerce'
)
```

```python

# 4. Verify the conversions
print("Cleaned data shape:", df_clean.shape)

print("\nNew column data types:")
print(df_clean[
    ['Engagement_Rate_Num',
     'Recalculated_Engagement_Rate_Num',
     'CTR_Percentage',
     'Post_Date_Clean']
].dtypes)

print("\nMissing values in new columns:")
print(df_clean[
    ['Engagement_Rate_Num',
     'Recalculated_Engagement_Rate_Num',
     'CTR_Percentage',
     'Post_Date_Clean']
].isnull().sum())

print("\nSample converted values:")
print(df_clean[
    ['Engagement_Rate', 'Engagement_Rate_Num',
     'Click_Through_Rate', 'CTR_Percentage',
     'Post_Date', 'Post_Date_Clean']
].head(10))
```

```python

# Ensure month is available for trend analysis
df_clean['Month_Label'] = (
    df_clean['Post_Date_Clean']
    .dt.to_period('M')
    .astype(str)
)

# Chart 1: Average engagement rate by platform
platform_data = (
    df_clean.groupby('Platform')['Recalculated_Engagement_Rate_Num']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
platform_data.plot(kind='bar')
plt.title('Average Engagement Rate by Platform')
plt.xlabel('Platform')
plt.ylabel('Average Engagement Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Average views by content category
category_data = (
    df_clean.groupby('Content_Category')['Views']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))
category_data.plot(kind='bar')
plt.title('Average Views by Content Category')
plt.xlabel('Content Category')
plt.ylabel('Average Views')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Chart 3: Organic vs Sponsored average views
content_data = (
    df_clean.groupby('Content_Type')['Views']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(7, 5))
content_data.plot(kind='bar')
plt.title('Average Views: Organic vs Sponsored')
plt.xlabel('Content Type')
plt.ylabel('Average Views')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Chart 4: Monthly average engagement trend
monthly_data = (
    df_clean.groupby('Month_Label')['Engagement']
    .mean()
    .sort_index()
)

plt.figure(figsize=(12, 5))
monthly_data.plot(kind='line', marker='o')
plt.title('Monthly Average Engagement Trend')
plt.xlabel('Month')
plt.ylabel('Average Engagement')
plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
# Chart 5: Average engagement by region
region_data = (
    df_clean.groupby('Region')['Engagement']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
region_data.plot(kind='bar')
plt.title('Average Engagement by Region')
plt.xlabel('Region')
plt.ylabel('Average Engagement')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 1. Overall KPIs
print("TOTAL POSTS:", len(df_clean))
print("TOTAL VIEWS:", f"{df_clean['Views'].sum():,}")
print(
    "AVERAGE ENGAGEMENT RATE:",
    round(df_clean['Recalculated_Engagement_Rate_Num'].mean(), 2), "%"
)

```

```python
# 2. Best platform by average engagement rate
platform_performance = (
    df_clean.groupby('Platform')
    .agg(
        total_posts=('Post_ID', 'size'),
        avg_views=('Views', 'mean'),
        avg_engagement=('Engagement', 'mean'),
        avg_engagement_rate=('Recalculated_Engagement_Rate_Num', 'mean')
    )
    .round(2)
    .sort_values('avg_engagement_rate', ascending=False)
)

print("\nPLATFORM PERFORMANCE")
display(platform_performance)
```

```python
# 3. Best content category by average views
category_performance = (
    df_clean.groupby('Content_Category')
    .agg(
        total_posts=('Post_ID', 'size'),
        avg_views=('Views', 'mean'),
        avg_engagement=('Engagement', 'mean')
    )
    .round(2)
    .sort_values('avg_views', ascending=False)
)

print("\nCONTENT CATEGORY PERFORMANCE")
display(category_performance)
```

```python
# 4. Organic vs Sponsored
content_performance = (
    df_clean.groupby('Content_Type')
    .agg(
        total_posts=('Post_ID', 'size'),
        avg_views=('Views', 'mean'),
        avg_engagement=('Engagement', 'mean'),
        avg_engagement_rate=('Recalculated_Engagement_Rate_Num', 'mean')
    )
    .round(2)
)

print("\nORGANIC VS SPONSORED")
display(content_performance)
```

```python
# 5. Best region by average engagement
region_performance = (
    df_clean.groupby('Region')
    .agg(
        total_posts=('Post_ID', 'size'),
        avg_engagement=('Engagement', 'mean'),
        avg_views=('Views', 'mean')
    )
    .round(2)
    .sort_values('avg_engagement', ascending=False)
)

print("\nREGION PERFORMANCE")
display(region_performance)
```
